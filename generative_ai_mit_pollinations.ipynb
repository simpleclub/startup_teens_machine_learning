# Generative KI mit Pollinations.AI

## Einführung in Generative KI

Generative KI ist ein Teilbereich der künstlichen Intelligenz, der darauf abzielt, neue Inhalte zu erstellen, die den Trainingsdaten ähneln. Im Gegensatz zu diskriminativen Modellen, die Daten klassifizieren oder vorhersagen, können generative Modelle neue Bilder, Texte oder Musik erzeugen.

In diesem Notebook werden wir Pollinations.AI verwenden, einen kostenlosen Dienst ohne Anmeldung, um Bilder mit Hilfe von generativen KI-Modellen zu erstellen.

## Benötigte Bibliotheken installieren

Wir benötigen nur einige grundlegende Python-Bibliotheken, um mit Pollinations.AI zu arbeiten:

```python
# Bibliotheken installieren (nur in Google Colab notwendig)
# !pip install requests pillow matplotlib
```

```python
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
```

## Funktion zum Generieren von Bildern

Wir erstellen eine Funktion, die die Pollinations.AI API verwendet, um Bilder basierend auf einer Textbeschreibung (Prompt) zu generieren:

```python
def generiere_bild(prompt, breite=512, hoehe=512, modell="stable-diffusion-v1-5"):
    """
    Generiert ein Bild mit Pollinations.AI API
    
    Parameter:
        prompt (str): Textbeschreibung des zu generierenden Bildes
        breite (int): Breite des Ausgabebildes (Standard: 512)
        hoehe (int): Höhe des Ausgabebildes (Standard: 512)
        modell (str): Zu verwendendes Modell (Standard: stable-diffusion-v1-5)
                      Andere Optionen: "sdxl", "kandinsky", "dalle-mini"
    
    Rückgabe:
        PIL.Image: Generiertes Bild
    """
    # Pollinations.AI API-Endpunkt
    api_url = "https://image.pollinations.ai/prompt/"
    
    # URL für die Anfrage erstellen
    # Format: {api_url}/{prompt}?width={breite}&height={hoehe}&model={modell}
    full_url = f"{api_url}{prompt}?width={breite}&height={hoehe}&model={modell}"
    
    # Anfrage senden
    response = requests.get(full_url)
    
    if response.status_code == 200:
        # Antwortinhalt in ein Bild umwandeln
        bild = Image.open(BytesIO(response.content))
        return bild
    else:
        print(f"Fehler: {response.status_code} - {response.text}")
        return None
```

## Beispiel 1: Ein einfaches Bild generieren

Lass uns ein Bild einer futuristischen Stadt generieren:

```python
# Prompt definieren
prompt = "Eine futuristische Stadt mit fliegenden Autos und Neonlichtern"

# Bild generieren
bild = generiere_bild(prompt)

# Bild anzeigen
if bild:
    plt.figure(figsize=(10, 10))
    plt.imshow(bild)
    plt.axis('off')
    plt.title(prompt)
    plt.show()
```

## Beispiel 2: Verschiedene Modelle verwenden

Pollinations.AI bietet verschiedene Modelle an. Lass uns das SDXL-Modell für eine bessere Qualität verwenden:

```python
# Prompt definieren
prompt = "Ein Roboter, der in einem Klassenzimmer Mathematik unterrichtet"

# Bild mit SDXL-Modell generieren
bild = generiere_bild(
    prompt=prompt,
    breite=768,
    hoehe=512,
    modell="sdxl"
)

# Bild anzeigen
if bild:
    plt.figure(figsize=(12, 8))
    plt.imshow(bild)
    plt.axis('off')
    plt.title(prompt + " (SDXL-Modell)")
    plt.show()
```

## Beispiel 3: Mehrere Bilder vergleichen

Wir können verschiedene Modelle vergleichen, indem wir mehrere Bilder mit demselben Prompt generieren:

```python
# Prompt definieren
prompt = "Ein Hund spielt im Park mit einem Ball"

# Verschiedene Modelle definieren
modelle = ["stable-diffusion-v1-5", "sdxl", "kandinsky"]

# Bilder mit verschiedenen Modellen generieren
bilder = {}
for modell in modelle:
    bilder[modell] = generiere_bild(prompt=prompt, modell=modell)

# Bilder nebeneinander anzeigen
plt.figure(figsize=(15, 5))
for i, (modell, bild) in enumerate(bilder.items()):
    if bild:
        plt.subplot(1, 3, i+1)
        plt.imshow(bild)
        plt.axis('off')
        plt.title(f"Modell: {modell}")
plt.suptitle(prompt)
plt.tight_layout()
plt.show()
```

## Beispiel 4: Bilder speichern

Wir können die generierten Bilder auch auf der Festplatte speichern:

```python
# Prompt definieren
prompt = "Eine Landschaft mit Bergen und einem See bei Sonnenuntergang"

# Bild generieren
bild = generiere_bild(
    prompt=prompt,
    breite=768,
    hoehe=512,
    modell="sdxl"
)

# Bild anzeigen
if bild:
    plt.figure(figsize=(12, 8))
    plt.imshow(bild)
    plt.axis('off')
    plt.title(prompt)
    plt.show()
    
    # Bild speichern
    bild.save("landschaft.png")
    print("Bild wurde als 'landschaft.png' gespeichert.")
```
